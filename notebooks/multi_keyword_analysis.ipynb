# Multi-Keyword Google Search Trends Analysis - India 2026

## Introduction

This analysis compares Google search interest for multiple technology and AI-related keywords in India during 2026.

The keywords include ChatGPT, Google Gemini, Claude, Python, and Data Science.

The objective is to compare search-interest patterns and identify differences in how interest in these topics changed over time.

## 1. Importing Libraries

```python
import pandas as pd
import matplotlib.pyplot as plt
```

## 2. Loading the Dataset

```python
multi_df = pd.read_csv("../data/raw/multi_keyword_india_2026.csv")

multi_df.head()
```

## 3. Understanding the Dataset

```python
multi_df.shape
```

```python
multi_df.columns
```

```python
multi_df.head(10)
```

```python
multi_df.info()
```

```python
multi_df.isnull().sum()
```

## 4. Data Cleaning

```python
multi_df = multi_df.rename(columns={
    "chatgpt": "ChatGPT",
    "google gemini": "Google Gemini",
    "claude": "Claude",
    "python": "Python",
    "data science": "Data Science"
})

multi_df["Time"] = pd.to_datetime(multi_df["Time"])

multi_df.head()
```

```python
multi_df.columns
```

```python
print("Start date:", multi_df["Time"].min())
print("End date:", multi_df["Time"].max())
```

## 5. Comparing Search Interest Across Keywords

This visualization compares Google search interest for ChatGPT, Google Gemini, Claude, Python, and Data Science in India during 2026.

```python
plt.figure(figsize=(12, 6))

plt.plot(multi_df["Time"], multi_df["ChatGPT"], marker="o", label="ChatGPT")
plt.plot(multi_df["Time"], multi_df["Google Gemini"], marker="o", label="Google Gemini")
plt.plot(multi_df["Time"], multi_df["Claude"], marker="o", label="Claude")
plt.plot(multi_df["Time"], multi_df["Python"], marker="o", label="Python")
plt.plot(multi_df["Time"], multi_df["Data Science"], marker="o", label="Data Science")

plt.title("Google Search Interest Comparison in India - 2026")
plt.xlabel("Date")
plt.ylabel("Search Interest")
plt.legend()
plt.xticks(rotation=45)
plt.grid()
plt.tight_layout()

plt.show()
```

## 6. Average Search Interest by Keyword

This section calculates the average Google search interest for each keyword during the selected period.

```python
average_interest = multi_df[
    ["ChatGPT", "Google Gemini", "Claude", "Python", "Data Science"]
].mean()

print(average_interest.round(2))
```

## 7. Average Search Interest Comparison

This bar chart compares the average Google search interest of the selected keywords in India during 2026.

```python
plt.figure(figsize=(10, 6))

average_interest.plot(kind="bar")

plt.title("Average Google Search Interest by Keyword - India 2026")
plt.xlabel("Keyword")
plt.ylabel("Average Search Interest")
plt.xticks(rotation=45)
plt.grid(axis="y")
plt.tight_layout()

plt.show()
```

## 8. Peak Search Interest by Keyword

This section identifies the highest recorded Google search-interest value and the corresponding date for each keyword.

```python
keywords = [
    "ChatGPT",
    "Google Gemini",
    "Claude",
    "Python",
    "Data Science"
]

for keyword in keywords:
    peak_index = multi_df[keyword].idxmax()
    
    print(keyword)
    print("Peak date:", multi_df.loc[peak_index, "Time"])
    print("Peak interest:", multi_df.loc[peak_index, keyword])
    print()
```

## 9. Peak Search Interest Comparison

This chart compares the maximum Google search-interest value reached by each keyword during the selected period.

```python
peak_interest = {}

for keyword in keywords:
    peak_interest[keyword] = multi_df[keyword].max()

peak_interest = pd.Series(peak_interest)

print(peak_interest)
```

```python
plt.figure(figsize=(10, 6))

peak_interest.plot(kind="bar")

plt.title("Peak Google Search Interest by Keyword - India 2026")
plt.xlabel("Keyword")
plt.ylabel("Peak Search Interest")
plt.xticks(rotation=45)
plt.grid(axis="y")
plt.tight_layout()

plt.show()
```

## 10. Changes in Search Interest

This section identifies the week-to-week changes in search interest for each keyword. It helps identify periods of sudden increases or decreases in search interest.

```python
change_df = multi_df.copy()

for keyword in keywords:
    change_df[f"{keyword}_Change"] = change_df[keyword].diff()

change_df.head()
```

```python
for keyword in keywords:
    change_column = f"{keyword}_Change"
    index = change_df[change_column].idxmax()

    print(keyword)
    print("Date:", change_df.loc[index, "Time"])
    print("Increase:", round(change_df.loc[index, change_column], 2))
    print()
```

```python
for keyword in keywords:
    change_column = f"{keyword}_Change"
    index = change_df[change_column].idxmin()

    print(keyword)
    print("Date:", change_df.loc[index, "Time"])
    print("Decrease:", round(change_df.loc[index, change_column], 2))
    print()
```

## 11. Largest Weekly Changes by Keyword

This section compares the largest week-to-week increases and decreases in Google search interest for each keyword.

```python
largest_increases = {}
largest_decreases = {}

for keyword in keywords:
    change_column = f"{keyword}_Change"
    
    largest_increases[keyword] = change_df[change_column].max()
    largest_decreases[keyword] = change_df[change_column].min()

increase_series = pd.Series(largest_increases)
decrease_series = pd.Series(largest_decreases)

print("Largest increases:")
print(increase_series)

print("\nLargest decreases:")
print(decrease_series)
```

```python
plt.figure(figsize=(10, 6))

x = range(len(keywords))

plt.bar(
    [i - 0.2 for i in x],
    increase_series.values,
    width=0.4,
    label="Largest Increase"
)

plt.bar(
    [i + 0.2 for i in x],
    decrease_series.values,
    width=0.4,
    label="Largest Decrease"
)

plt.xticks(x, keywords, rotation=45)
plt.xlabel("Keyword")
plt.ylabel("Change in Search Interest")
plt.title("Largest Weekly Changes in Search Interest - India 2026")
plt.legend()
plt.grid(axis="y")
plt.tight_layout()

plt.show()
```

## 12. Monthly Search Interest Analysis

This section analyzes the average Google search interest for each keyword across the months of 2026.

```python
multi_df["Month"] = multi_df["Time"].dt.month
multi_df["Month_Name"] = multi_df["Time"].dt.month_name()

multi_df.head()
```

```python
month_order = [
    "January",
    "February",
    "March",
    "April",
    "May",
    "June",
    "July",
    "August",
    "September",
    "October",
    "November",
    "December"
]

multi_df["Month_Name"] = pd.Categorical(
    multi_df["Month_Name"],
    categories=month_order,
    ordered=True
)
```

```python
monthly_average = (
    multi_df
    .groupby("Month_Name", observed=False)[keywords]
    .mean()
)

monthly_average
```

```python
monthly_average.round(2)
```

## 13. Monthly Search Interest Comparison

This visualization compares the average monthly Google search interest for all five keywords in India during 2026.

```python
plt.figure(figsize=(12, 6))

for keyword in keywords:
    plt.plot(
        monthly_average.index,
        monthly_average[keyword],
        marker="o",
        label=keyword
    )

plt.title("Monthly Google Search Interest Comparison - India 2026")
plt.xlabel("Month")
plt.ylabel("Average Search Interest")
plt.legend()
plt.xticks(rotation=45)
plt.grid()
plt.tight_layout()
plt.show()
```

## 14. Trends and Patterns

The monthly search-interest data was analyzed to identify major trends and changes in search behavior across the five keywords.

```python
# Calculate percentage change from January to September

for keyword in keywords:
    january_value = monthly_average.loc["January", keyword]
    september_value = monthly_average.loc["September", keyword]

    percentage_change = (
        (september_value - january_value) / january_value
    ) * 100

    print(
        keyword,
        "percentage change from January to September:",
        round(percentage_change, 2), "%"
    )
```

## 15. Correlation Analysis

Correlation analysis is used to measure the relationship between the search-interest patterns of different keywords.

A correlation value close to +1 indicates that two keywords tend to move in the same direction, while a value close to -1 indicates that they tend to move in opposite directions.

```python
correlation_matrix = multi_df[keywords].corr()

correlation_matrix.round(2)
```

```python
import seaborn as sns

plt.figure(figsize=(10, 6))

sns.heatmap(
    correlation_matrix,
    annot=True,
    cmap="coolwarm",
    fmt=".2f"
)

plt.title("Correlation of Google Search Interest - India 2026")
plt.tight_layout()
plt.show()
```

```python
variable_keywords = [
    "ChatGPT",
    "Google Gemini",
    "Claude",
    "Python"
]

correlation_matrix = multi_df[variable_keywords].corr()

plt.figure(figsize=(8, 6))

sns.heatmap(
    correlation_matrix,
    annot=True,
    cmap="coolwarm",
    fmt=".2f"
)

plt.title("Correlation of Google Search Interest - India 2026")
plt.tight_layout()
plt.show()
```
